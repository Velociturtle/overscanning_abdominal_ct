```python
#!/usr/bin/env python3
import os
import shutil
import glob
import dicom2nifti
import dicom2nifti.settings as settings

# ─── CONFIG ─────────────────────────────────────────────────────────────────────
DICOM_ROOT   = r"D:\Abdomen_CT_Bone_Mets"
NIFTI_ROOT   = r"D:\Abdomen_CT_Bone_Mets_Nifti"

# ─── ENABLE REORIENTATION ───────────────────────────────────────────────────────
settings.reorient_nifti = True   # force correct LPS→RAS orientation

# ─── PROCESS EACH PATIENT SUBFOLDER ─────────────────────────────────────────────
for pid in os.listdir(DICOM_ROOT):
    src_dir = os.path.join(DICOM_ROOT, pid)
    if not os.path.isdir(src_dir):
        continue

    out_dir = os.path.join(NIFTI_ROOT, pid)
    # 1) Remove old output
    if os.path.isdir(out_dir):
        print(f"Deleting old output: {out_dir}")
        shutil.rmtree(out_dir)
    os.makedirs(out_dir, exist_ok=True)

    # 2) Convert directory
    print(f"\nConverting {pid} → {out_dir} …")
    try:
        dicom2nifti.convert_directory(src_dir, out_dir, compression=True)
    except Exception as e:
        print(f"⚠️  Conversion failed for {pid}: {e}")
        continue

    # 3) Find and rename the single NIfTI
    nii_files = glob.glob(os.path.join(out_dir, "*.nii*"))
    if len(nii_files) == 1:
        src_nii = nii_files[0]
        dst_nii = os.path.join(out_dir, f"{pid}.nii.gz")
        os.rename(src_nii, dst_nii)
        print(f"  → Renamed to {dst_nii}")
    else:
        print(f"⚠️  Unexpected files in {out_dir}: {nii_files}")

print("\nBatch conversion complete. All outputs are reoriented head→feet.")  
```

```python
#!/usr/bin/env python3
import os
import glob

# ─── CONFIG ─────────────────────────────────────────────────────────────────────
NIFTI_ROOT = r"D:\Abdomen_CT_Bone_Mets_Nifti"

# ─── FIND BLANK FOLDERS ─────────────────────────────────────────────────────────
blank_folders = []

for pid in os.listdir(NIFTI_ROOT):
    folder = os.path.join(NIFTI_ROOT, pid)
    if not os.path.isdir(folder):
        continue
    # look for any .nii or .nii.gz files
    nifti_files = glob.glob(os.path.join(folder, "*.nii")) + \
                  glob.glob(os.path.join(folder, "*.nii.gz"))
    if len(nifti_files) == 0:
        blank_folders.append(folder)

# ─── OUTPUT RESULTS ─────────────────────────────────────────────────────────────
if blank_folders:
    print("Folders with no NIfTI output:")
    for f in blank_folders:
        print(f"  {f}")
else:
    print("All patient folders contain at least one NIfTI file.")
```
