```python
from pathlib import Path

def core_no_ext(p: Path) -> str:
    """Return filename with *all* extensions (.nii / .nii.gz) removed."""
    name = p.name
    if name.lower().endswith(".nii.gz"):
        return name[:-7]          # strip ".nii.gz"
    if name.lower().endswith(".nii"):
        return name[:-4]          # strip ".nii"
    return name

root = Path(r"D:\Abdomen_CT_Bone_Mets_Nifti")

issues       = []   # folders with ≠ 1 *.nii*
roi1_details = []   # folders containing ROI1 files
roi1_total   = 0

for folder in [d for d in root.iterdir() if d.is_dir()]:
    nii_files  = list(folder.glob("*.nii*"))                # .nii or .nii.gz
    roi1_files = [f for f in nii_files if core_no_ext(f).endswith("_ROI1")]
    roi1_count = len(roi1_files)
    roi1_total += roi1_count

    if len(nii_files) != 1:
        issues.append((folder.name, len(nii_files)))

    if roi1_count:
        roi1_details.append((folder.name, roi1_count))

# ── Results ────────────────────────────────────────────────────────────────────
if issues:
    print("❌  Folders with the wrong number of NIfTI files:")
    for name, count in issues:
        print(f"   {name:<30} → {count} total NIfTI files")
else:
    print("✅  Every sub‑folder contains exactly one NIfTI file.")

if roi1_details:
    print("\n📌  ROI1‑labelled NIfTI files detected:")
    for name, count in roi1_details:
        print(f"   {name:<30} → {count} ROI1 file(s)")
    print(f"\nTotal ROI1 files across all folders: {roi1_total}")
else:
    print("\n✅  No ROI1‑labelled NIfTI files found.")
```

```python
from pathlib import Path

root = Path(r"D:\Abdomen_CT_Bone_Mets_Nifti")
deleted = 0

# delete any file whose *base name* (minus .nii / .nii.gz) ends with “_ROI1”
for p in root.rglob("*.nii*"):
    name = p.name.lower()
    base = name[:-7] if name.endswith(".nii.gz") else name[:-4] if name.endswith(".nii") else name
    if base.endswith("_roi1"):
        try:
            p.unlink()
            deleted += 1
            print(f"🗑️  Deleted {p}")
        except Exception as e:
            print(f"⚠️  Couldn’t delete {p} → {e}")

print(f"\n✅  Finished. {deleted} ROI1 file(s) removed.")
```
